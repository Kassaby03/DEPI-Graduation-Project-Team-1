```python
def temp_cat(temp):
    if temp < 10:
        return 'Cold'
    elif temp < 20:
        return 'Mild'
    elif temp < 30:
        return 'Warm'
    else:
        return 'Hot'

FactWeatherObservation['Temp_category'] = FactWeatherObservation['temperature'].apply(temp_cat)
```

```python
def cloud_level(value):
    if value <= 25:
        return 'Clear'
    elif value <= 50:
        return 'Partly Cloudy'
    elif value <= 75:
        return 'Cloudy'
    else:
        return 'Overcast'

FactWeatherObservation['Cloud_level'] = FactWeatherObservation['cloud_coverage'].apply(cloud_level)
```

```python
FactWeatherObservation['Cloud_level'].value_counts()
```

```python
def visibility_level(v):
    if v <= 1000:
        return 'Very Low'
    elif v <= 5000:
        return 'Low'
    elif v <= 9000:
        return 'Moderate'
    else:
        return 'High'

FactWeatherObservation['Visibility_level'] = FactWeatherObservation['visibility'].apply(visibility_level)
```

```python
def wind_category(speed):
    if speed <= 2.5:
        return 'Calm'
    elif speed <= 5.0:
        return 'Light Breeze'
    elif speed <= 10.0:
        return 'Moderate'
    elif speed <= 20.0:
        return 'Strong'
    else:
        return 'Stormy'

FactWeatherObservation['Wind_category'] = FactWeatherObservation['wind_speed'].apply(wind_category)
```

```python
FactWeatherObservation['Comfort_index'] = (
    FactWeatherObservation['temperature'] - (0.55 - 0.0055 * FactWeatherObservation['humidity']) * (FactWeatherObservation['temperature'] - 14.5)
).round(2)
```

```python
FactWeatherObservation.sample(10)
```

```python
def comfort_label(value):
    if value < 15:
        return "Very Cold"
    elif 15 <= value <= 22:
        return "Comfortable"
    elif 23 <= value <= 28:
        return "Warm"
    elif 29 <= value <= 35:
        return "Hot"
    else:
        return "Very Hot"

FactWeatherObservation['Comfort_level'] = FactWeatherObservation['Comfort_index'].apply(comfort_label)
```
